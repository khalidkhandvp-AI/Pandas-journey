```python
#Making one dataframe from many using MEJGE having same name columns
a=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Khalid","khan","MKK"]
})
b=pd.DataFrame({
    "ID":[1,2,3],
    "Marks":[99,98,97]
})
c=pd.merge(a,b,on="ID")
print(c)
```

```python
#Making one dataframe from many using MEJGE having different name columns
a=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Khalid","khan","MKK"]
})
b=pd.DataFrame({
    "IDS":[1,2,3],
    "Marks":[99,98,97]
})
c=pd.merge(a,b,left_on="ID",right_on="IDS")
print(c)
```

```python
#Join using in the code
a=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Khalid","khan","MKK"]
},index=[1,2,3])
b=pd.DataFrame({
    "Marks":[99,98,97]
},index=[1,2,4])
a.join(b,how="inner")
a.join(b,how="left")
a.join(b,how="right")
a.join(b,how="outer")
```

```python
#Concatenation of dataframes
a=pd.DataFrame({
    "ID":[1,2,3],
    "Name":["Khalid","khan","MKK"]
})
b=pd.DataFrame({
    "IDS":[1,2,3],
    "Marks":[99,98,97]
})
print(pd.concat([a,b]))
print(pd.concat([a,b],axis=0))
print(pd.concat([a,b],axis=1))
```
