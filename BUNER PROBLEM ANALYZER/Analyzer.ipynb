```python
import pandas as pd 
df=pd.read_csv("Villages_data.csv")
print(df)
```

```python
#Analysis of all the problems
print(df["Road"].value_counts())
print(df["Electricity"].value_counts())
print(df["Hospital"].value_counts())
print(df["Education"].value_counts())
```

```python
#giving ranks to the problems
road_map={
    "bad":1,
    "medium":2,
    "good":3,
}
electricity_map={
    "bad":1,
    "medium":2,
    "medium+":2.5,
    "good":3
}
hospital_map={
    "NO":1,
    "YES":2,
    "A LOT":3
}
education_map={
    "bad":1,
    "medium":2,
    "good":3
}
```

```python
#making rank colums
df["Road_score"]=df["Road"].map(road_map)
df["Electricity_score"]=df["Electricity"].map(electricity_map)
df["hospital_score"]=df["Hospital"].map(hospital_map)
df["Education_score"]=df["Education"].map(education_map)
print(df)
```

```python
#adding all the scores
df["Total_Score"]=(df["Road_score"]+df["Electricity_score"]+df["hospital_score"]+df["Education_score"])
```

```python
print(df[["Name","Total_Score"]])
```

```python
#The best and worst villages in Buner
best_village=df.loc[df["Total_Score"].idxmax()]
worst_village=df.loc[df["Total_Score"].idxmin()]
best=best_village["Name"]
worst=worst_village["Name"]
```

```python
#printing the names of the best and worst villages
print(f"According to this repost the best village in Buner is {best} and the worst village is {worst}")
```
